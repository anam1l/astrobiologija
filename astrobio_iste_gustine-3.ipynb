```python
def bin(wvlngth_list, flux_list, n_bins):
    wvl_bin = []
    flx_bin = []
    flux_time = []  #srednji flux po binovima kroz vreme
    wvl_time = [] # srednja talasna duzina ..
    bin_edge = []

    for wvl, flx in zip(wvlngth_list, flux_list):
        wvl = np.array(wvl)
        flx = np.array(flx)
        
        bin_size = len(flx) // n_bins
        
        # slice za binove
        wvl_bins = [wvl[(5-i)*bin_size:((5-i)+1)*bin_size] for i in range(n_bins)]
        flx_bins = [flx[(5-i)*bin_size:((5-i)+1)*bin_size] for i in range(n_bins)]
        
        
        # srednji flux po binovima
        mean_flux = [np.mean(fb) for fb in flx_bins]
        mean_wvl = [np.mean(wb) for wb in wvl_bins]

        edges = [wb[0] for wb in wvl_bins]   
        edges.append(wvl_bins[-1][-1])
        bin_edge.append(edges)
        
        wvl_bin.append(wvl_bins)
        flx_bin.append(flx_bins)
        flux_time.append(mean_flux)
        wvl_time.append(mean_wvl)
    
    return wvl_bin, flx_bin, flux_time, wvl_time,bin_edge

def binovanje(wvlngth_list, flux_list, n_bins):

    bin_edges = np.linspace(np.min(wvlngth_list[6]),np.max(wvlngth_list),n_bins + 1)
    

    flux_time = [] # srednje vrednosti binova
    wvl_time = []

    flux_bins_all = [] #sve vrednosti po binovima
    wvl_bins_all = []

    for wvl, flx in zip(wvlngth_list, flux_list):

        wvl = np.array(wvl)
        flx = np.array(flx)

        flux_bins = []
        wvl_bins = []

        mean_flux = []
        mean_wvl = []

        for i in range(n_bins):

            mask = (wvl >= bin_edges[i]) & (wvl < bin_edges[i+1])

            flux_in_bin = flx[mask]
            wvl_in_bin = wvl[mask]

            flux_bins.append(flux_in_bin)
            wvl_bins.append(wvl_in_bin)

            mean_flux.append(np.mean(flux_in_bin))
            mean_wvl.append(np.mean(wvl_in_bin))

        flux_bins_all.append(flux_bins)
        wvl_bins_all.append(wvl_bins)
        flux_time.append(mean_flux)
        wvl_time.append(mean_wvl)

    return (flux_bins_all,wvl_bins_all,np.array(flux_time),np.array(wvl_time),bin_edges)
```

```python
def periods_plot_data(lcID, data, ngrid,
                      peakHeight=0.6,
                      prominence=0.7,
                      minfq=None,
                      maxfq=None):
    
    hh1 = np.rot90(data).T / np.rot90(data).T.max()
    hh1arr = np.rot90(hh1.T)
    hh1arr1 = np.abs(hh1arr).sum(1) / np.abs(hh1arr).sum(1).max()

    fmin = 1 / minfq
    fmax = 1 / maxfq
    df = (fmax - fmin) / ngrid

    osax = np.arange(start=fmin, stop=fmax + df, step=df)
    xax = np.arange(start=fmin, stop=fmax + df, step=df / 2)

    from scipy import interpolate
    f = interpolate.interp1d(osax, np.abs(hh1arr1), fill_value="extrapolate")
    yax = f(xax)

    peaks, props = find_peaks(yax, peakHeight, prominence=prominence)

    return {
        "lcID": lcID,
        "xax": xax,
        "yax": np.abs(yax),
        "peaks": peaks,
        "props":props
    }
```

### Kendall's W

```python
def kendall_w(data):
    
    n = 10
    k = 6
    
    ranks = np.array([stats.rankdata(data[:, i]) for i in range(k)]).T
    
    R = np.sum(ranks, axis=1)
    R_mean = np.mean(R)
    S = np.sum((R - R_mean)**2)
    W = (12 * S) / (k**2 * (n**3 - n))
    return W

w_coeff = kendall_w(flux_time1)
```
